```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
df = df[['Tweet', 'Label']]

df.head()
```

```python
df['Label'].value_counts()
```

```python
class_counts = df['Label'].value_counts().reset_index()

class_counts.columns = ['Emotion', 'Count']

class_counts
```

```python
X = df['Tweet']
y = df['Label']
```

```python
label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print(label_encoder.classes_)
```

```python
print(y_encoded[:10])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)
```

Why did you use stratify?"

Answer:

"Because the dataset is imbalanced. Stratification ensures that approximately the same proportion of every emotion class is maintained in both the training and testing sets."

```python
tokenizer = Tokenizer()

tokenizer.fit_on_texts(X_train)
```

```python
X_train_encoded = tokenizer.texts_to_matrix(X_train, mode='binary')
X_test_encoded = tokenizer.texts_to_matrix(X_test, mode='binary')

print("Training shape:", X_train_encoded.shape)
print("Testing shape:", X_test_encoded.shape)

print("Vocabulary size:", len(tokenizer.word_index))
```

```python
X_train_encoded.shape
input_dim = 10000

input_dim = X_train_encoded.shape[1]

print("Input dimension:", input_dim)
```

```python
model = Sequential()

model.add(Dense(64, activation='relu', input_dim=input_dim))
model.add(Dense(32, activation='relu'))
model.add(Dense(6, activation='softmax'))
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

optimizer='adam'
```

```python
loss='sparse_categorical_crossentropy'


history = model.fit(
    X_train_encoded,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.1
)
```

```python
history = model.fit(
    X_train_encoded,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.1
)
```

```python
test_loss, test_accuracy = model.evaluate(
    X_test_encoded,
    y_test
)

print("Test Accuracy:", test_accuracy)
```

```python
train_loss, train_accuracy = model.evaluate(
    X_train_encoded,
    y_train,
    verbose=0
)

test_loss, test_accuracy = model.evaluate(
    X_test_encoded,
    y_test,
    verbose=0
)

print("Training Accuracy:", train_accuracy)
print("Test Accuracy:", test_accuracy)
```
